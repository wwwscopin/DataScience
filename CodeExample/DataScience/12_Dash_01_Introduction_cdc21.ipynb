## Overview: What is Dash?

[Dash](https://dash.plotly.com/installation) is a framework (from the `plotly` team) for rapidly building analytic "apps" or dashboards using Python. It is influenced and inspired by the **Shiny** project in R. To get a sense of what Dash can do, check out the [gallery here.](https://dash-gallery.plotly.host/Portal/)

As you can see, Dash incorporates widgets, charts and layouts within the framework of an in-browser app. 

From the docs: 

> Dash is a productive Python framework for building web analytic applications.  
> Written on top of Flask, Plotly.js, and React.js, Dash is ideal for building data visualization apps with highly custom user interfaces in pure Python. It's particularly suited for anyone who works with data in Python.  https://dash.plotly.com/introduction

The resulting application is viewable in your web browser. You don't need to know the full stack of Flask and React - you can simply write Python code using the high level Dash interface.

## Structure of a Dash application folder

The typical structure of a Dash application folder looks like this: 

>Application folder:  
---assets/  
---data/  
---app.py  
---README.MD  
---requirements.txt


Here is a description of each component:

* assets/   
    * This includes CSS files, images, fonts, etc  
* data/  
    * This includes data for the application. 
* app.py  
    * This is the Python script that is run when the application is launched. This is the main 'thing' in the folder. The layout of the application, data transformations, plots, etc will all be defined here. 
* README.MD  
    * This is a markdown file describing the application. 
* requirements.txt  
    * This is a list of Python packages (modules) and versions that are required to run the application. This is typically generated with pip, and then a pip command can be used to install all the required packages. usual practice would be to establish a new virtual environment, then install the required packages from this file, than launch the application. We will not include this here, but you will often see this in applications shared online. 


We have included a template of this structure at this location: `Apps/00_template/`  

When you create a new application, you can copy this template folder and then rename the copy to make it easier to generate a new app.

The main parts of the Dash app are the following:  
* prior code
    + module imports, data setup, and any global variable setup occur here
* app.layout = ...

## More about HTML components

The `dash_html_components` contains all the html tags you may be familiar with. 

Each html component class has standard keyword arguments for html:

* `children=`
    + You can pass in string, number, a single component, or a list of components.
* `style=`
    + supply a dictionary here to map style attributes. This includes text color, font size, text alignment, background color, and much more
    + the keys are camelCased - for example, backgroundColor instead of background-color
    + see this link for more general info on style attributes in html: https://www.w3schools.com/html/html_styles.asp
    + You can apply style updates inline to components here, but if you want to apply global style changes (like background color for entire page), you should modify the CSS stylesheet and place is in assets/  
    + here is some more info on HTML styles and CSS stylesheets: https://www.w3schools.com/html/html_css.asp  
* `id=`
    + this is the name of the component and must be unique across all components. This will be used in the callbacks. 
* `className=`
    + This is the HTML `class` argument, which allows styling to be applied to classes. 
    
Check the documentation for other keywords available for individual html components.

## More about Dash Core Components

`dcc` stands for Dash core components. https://dash.plotly.com/dash-core-components

These are high level abstractions that provide us with functionality we expect in an analytic application. These are not pure HTML - they are generated with JavaScript, HTML, and CSS through the React.js library. 


This is how we will include plotly figures: with `dcc.Graph()`, and many other components. I will show you some in the application, though they will not do anything yet! 


Here is the example structure for a component, `dcc.Dropdown()`:

```
              dcc.Dropdown(
                  options=[
                      {'label':'Texas','value':'TX'},
                      {'label':'Florida','value':'FL'}
                ],
                value='TX')
```

Notice the options are passed as a list of dictionaries, then the value argument sets the default value. The user can change this, and later we will learn how to use this in reactive programming! For now, I will just demonstrate a few. 


The dropdown can also be multi value:

```
              dcc.Dropdown(
                  options=[
                      {'label':'Texas','value':'TX'},
                      {'label':'Florida','value':'FL'}
                ],
                value='TX',
                multi=True)
```

Here are some more examples of dcc components:

```
              dcc.RadioItems(
                  options=[
                      {'label':'Texas','value':'TX'},
                      {'label':'Florida','value':'FL'}
                ],
                value='TX')
```

```
              dcc.Checklist(
                  options=[
                      {'label':'Texas','value':'TX'},
                      {'label':'Florida','value':'FL'}
                ]

```

## Styling the html inline, with CSS, or with className

Now let's explore how to style your HTML code. You can style the HTML code in multiple ways:

* Directly in the component through the `style=` argument
* Through the CSS sheet
* Through the CSS sheet by identifying a class name. 

Here is an example of styling directly in the CSS style argument - make this modification in you code and see the result - provide a dictionary to the style argument. 

```
html.H1(children='Hello World!', 
        style={"color":"red"})`
```


This would make the individual component red in color. 

Alternatively, you could include this line of code in your CSS style sheet under the assets folder:

```
h1 {
  color: red;
}
```

And the style would be set to red.

> Try to modify your application app.py file, save the file, then see the result in your web browser.


The final thing to mention is you can identify the html component to be a specific class using className - then you can reference that className in your CSS. Your application code would look like this:


```
html.H1(children='Hello World!', 
        className="header-title")`
```


Then you would add this to your CSS:

```
.header-title {
  color:red
}
```

Any of these will work - but leveraging the CSS file is the most efficient.
