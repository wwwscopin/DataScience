#### Slicing (Indexing)

* We have seen this a few times, let’s formally address it.
* We can slice (index) sequence like objects similarly
 * Use `[]`. Python starts counting at 0, so `var[0]` is the first element
 * We can get a range (a slice) using `":"` example: `var[0:2]`
 * Slices are __start__ inclusive, __finish__ exclusive

```python
# Slicing: first create a list of numbers
seq1 = list(range(1, 11))

seq1
```

```python
## Different slicing techniques
# first and second
seq1[:2]
```

```python
# Third to the end
seq1[2:]
```

```python
# up to the 8 element
seq1[:8]
```

```python
# only the 8 element
seq1[7:8]
```

The `":"` has three arguments: *Start*, *stop*, *increment* 

So far, we are leaving the increment blank and therefore it is the default value of 1. 

*Start*/*Stop* can be negative, implying count from the end. We'll explore this below. *Increment* default value is 1, but it can be any integer (including negative values).

```python
# Slice stepping 
print("Starting sequence:", seq1)
```

```python
# first from the end
seq1[:9:2]
```

```python
# first to the 9th by 2
seq1[:9:2]
```

```python
# 10th to the first by negative 2
seq1[9:0:-2]
```

```python
# why does this not print anything?
seq1[-2:-5]
```

```python
# Because of the step! (increment)
seq1[-2:-5:-1]
```

```python
# How would you Reverse the entire sequence? 
seq1[::-1]
```

### Sets

A common question is - how do I get the unique values of a list? Let's say I have the list below, and I wish to get only the unique values back from the list. Is there a `unique()` method?

```python
# list
letters1 = list("abcdcbaa")
letters1
```

We will learn later in the material (with NumPy and Pandas) that there is a unique method/function. But, if we are using base Python objects (like lists and sets), there is not a unique method! Instead, we create a set, which is a new type of collection we have not yet learned about.

Sets share similarities to both lists and dictionaries. They differ from the former, in that they enforce uniqueness in the collection . They are essentially a dictionary consisting of only keys.

let's look at two lists of letters:

```python
letters1
```

```python
set1 = set(letters1)

set1
```

Notice the uniqueness enforced here.

The advantage of using the set is to perform set operations (anything you could represent as a Venn diagram). Let's take another set, with some letters in common:

```python
letters2 = list("cdeffeddfc")

letters2
```

```python
set2 = set(letters2)

set2
```

What are the letters that are in the first set, that are not in the other? We can use set operations to answer questions like this.

```python
# set operations - difference
set1 - set2
```

What is the unique combination of both sets?

```python
set1.union(set2)
```

What do both have in common?

```python
set1.intersection(set2)
```

Is everything in the first set contained in the second (a _subset_ )?

```python
set1.issubset(set2)
```

What about the letters from a - f?

```python
set1.issubset(set("abcdef"))
```

Set operations like this are an important aspect of a programming language. This is the method we use for set operations in base Python. We will see similar operations in other Python libraries later in the course.

## Control Flow

### If-Else flows

The `if` statement checks a condition. If true, it executes the first code block:  

```python
the_answer = 30

if (the_answer == 42):
    print("good job.")

    
print("not part of the if statement")
```

How does Python know what should be executed in the case that the `if` statement were true?

Note the indentation: **Whitespace is significant for all types of control flow in Python**

This is a deliberate design decision, intended to improve the readability of Python code. You can scan the margin of code and see where the flow of logic is shaped, or a function is defined.

Additional `if` conditions can be checked with `elif`. If none of these are true, the `else` block is executed.

```python
the_answer = 30

if the_answer == 42:
    print("good job.")
elif the_answer == 0:
    print("nice try.")
else:
    print("better luck next time.")
```

### While Loop

The while loop in Python works similarly to other languages. You start with a condition, once it evaluates to false the loop is terminated.  


```python
## While Loops
counter = 5

while counter > 0:
    print(f"Still Positive...i = {counter}")  # Note the f-string here
    counter = counter - 1
    print(counter)
```

### For Loops

The Python `for` loop looks almost like pseudo-code! 

Again, note the significance of whitespace to indicate the interior of the loop. Indentation matters! In fact, the loop contents are defined by having similarly indented code. 

```python
#### Control flow
## For Loops
for i in range(10):
    print(i)
    print('Next!')
```

We often want to enumerate within the for loop (establish a counter). You can do this using tuple unpacking and the `enumerate()` function. 

`enumerate()` returns something called a _generator_ that will yield a tuple of length two, with the values of the iterable and a counter. We haven't discussed generators yet...think of them like a lazy version of a collection. In this case, instead of making the entire tuple at once, `enumerate()` creates a generator that will give us the elements of the tuple as we call them.

```python
## Create a list of names
name_list = ['Evan', 'John', 'Kerry']

# use for loop to iterate through the list
for idx in name_list:
    print(idx)
```

```python
enumerate(name_list) # this is a generator
```

```python
## ask for elements of the generator:
for i in enumerate(name_list):
    i
```

We typically use tuple unpacking to catch the counter part like this:

```python
## enumeration within for loop
for counter, element in enumerate(["The", "Dog", "Cat"]):
    if counter == 0:
        print("First Word: " + element)
    else:
        print("Next Word: " + element)
```

You should use `enumerate()` to keep a counter in a for loop. Open the Python enumerate help document to view more information by submitting `?enumerate`

### List Comprehension

List comprehension is a popular part of Python. Essentially, it's a for loop that returns a list. If we didn't have list comprehension, we would need to first declare the new list, then add in the elements as we iterated through something. That would look like this:

```python
## Create a new variable that is x1 squared
x1 = list(range(10))

x1
```

```python
y1 = []

for i in x1:
    y1.append(i*i)

y1
```

Instead, we use list comprehension to accomplish this all at once. 

```python
## List comprehension
[i*i for i in range(10)]
```

We can add a filter like this:

```python
[i*i for i in range(10) if i > 5]
```

We can also add in a conditional return in the list comprehension:

```python
## If we want a conditional we move it up front
[i*i if i > 5 else i for i in range(10)]
```

* Functions in Python:

This is the general syntax to define a new function:

>`def new_func(par1,par2,some_other_par="default:"):
    """This function does this..."""
    _do_some_stuff_
    return some_related_stuff`
    
Let's try a simple function!

```python
# First Function
def mult_by_3(x, weight=1):
    """Multiplies a number, x, by 3
    
    Takes an optional weight parameter."""
    result = x*weight*3
    return result
```

```python
mult_by_3(5)
```

```python
mult_by_3(4)
```

```python
mult_by_3(4, 0.5)
```

```python
mult_by_3(weight=0.25, x=4)
```

```python
help(mult_by_3)
```

Functions can return multiple values (and also unpack them!)

```python
# Multiple Values
def multi_func(a, b, c):
    return (a + 1, b + 3, c - 3)
```

```python
a1, b1, c1 = multi_func(1, 1, 1)

a1
```

## Anonymous Functions

These are so called `"lambda"` functions.  
One line function expressions (useful construct in data science):  
>`new_func = lambda par [,par2,…]:  expression`

Let's try using a `lambda` function

```python
# Lambda
lam_func = lambda x1, x2: x1 + x2
```

```python
lam_func(3, 4)
```

```python
# Multiple returns
lam_func2 = lambda x1, x2: (x1 + x2, x1 - x2)

lam_func2(3, 4)
```
